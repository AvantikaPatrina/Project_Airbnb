```python
import spacy
import pandas as pd
import string
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

# Load SpaCy's English tokenizer
nlp = spacy.load('en_core_web_sm')

# Load the dataset
df = pd.read_csv("airbnb.csv")

# Initialize necessary tools
stop_words = set(stopwords.words('english'))
lemmatizer = WordNetLemmatizer()

# Function to clean a single description using SpaCy tokenizer
def clean_description(description):
    # Tokenize using spacy
    doc = nlp(description.lower())  # Tokenize and convert to lowercase
    cleaned_tokens = []

    for token in doc:
        # Remove punctuation and stopwords
        if token.text not in string.punctuation and token.text not in stop_words:
            # Lemmatize the token
            lemma = lemmatizer.lemmatize(token.text)
            cleaned_tokens.append(lemma)

    # Join tokens back into a single string
    return ' '.join(cleaned_tokens)

# Apply cleaning to all descriptions
df['cleaned_description'] = df['description'].fillna("").apply(clean_description)

# Preview cleaned descriptions
print(df['cleaned_description'].head())
```

```python
from sentence_transformers import SentenceTransformer

# Load a lightweight, fast, and accurate model
model = SentenceTransformer('all-MiniLM-L6-v2')
```

```python
len(df)
```

```python
# df['cleaned_description'] already exists
descriptions = df['cleaned_description'].fillna("").tolist()
```

```python
df.to_csv('lemmatized df.csv',index=False)
```

```python
# Get embeddings (this will return a NumPy array)
embeddings = model.encode(descriptions, show_progress_bar=True)
```

```python
embeddings.shape
```

```python
# Optional: add to dataframe
import numpy as np
df['embedding'] = list(embeddings)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

# Example: find top 3 listings similar to the first one
similarities = cosine_similarity([embeddings[0]], embeddings)[0]
top_indices = similarities.argsort()[-4:-1][::-1]  # top 3 excluding itself

df.iloc[top_indices][['description']]
```
